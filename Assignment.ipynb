```python
# 1. Counting Words in a File
def count_words(filename="document.txt"):
    with open(filename) as f: return len(f.read().split())



```

```python
def read_reverse(filename="reverse.txt"):
    with open(filename) as f: print(''.join(reversed(f.readlines())))
```

```python
def get_stats(filename="stats.txt"):
    with open(filename) as f:
        lines = f.readlines()
        return len(lines), len("".join(lines).split()), sum(len(l) for l in lines)
```
