```python
import numpy as np

arr = np.random.randint(1, 21, (5, 5))
arr[:, 2] = 1  # 3rd column = 1
print(arr)
```

```python
arr = np.arange(1, 17).reshape(4, 4)
np.fill_diagonal(arr, 0)
print(arr)
```

```python
arr = np.random.randint(1, 10, (5, 5))
border = np.r_[arr[0, :], arr[1:-1, -1], arr[-1, ::-1], arr[1:-1, 0]]
print(border)
```

```python
arr = np.arange(1, 17).reshape(4, 4)
print("Row sums:", arr.sum(axis=1))
print("Col sums:", arr.sum(axis=0))
```

```python
arr = np.arange(1, 10).reshape(3, 3)
print(arr.reshape(1, 9))
print(arr.reshape(9, 1))
```

```python
data = [('Alice', 28, 65.5), ('Bob', 22, 70.2), ('Charlie', 35, 78.0)]
arr = np.array(data, dtype=[('name', 'U10'), ('age', 'i4'), ('weight', 'f4')])
print(np.sort(arr, order='age'))
```
