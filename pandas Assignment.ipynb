```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C'])
df['Product'] = df['A'] * df['B']
print(df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (4, 3)), columns=['A', 'B', 'C'])
print(df)
print("\nRow sums:\n", df.sum(axis=1))
print("\nColumn sums:\n", df.sum(axis=0))
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C'])
df.iloc[1, 0], df.iloc[3, 2] = np.nan, np.nan  # Introduce NaNs

print("Original with NaNs:\n", df)
df = df.fillna(df.mean())
print("\nAfter handling NaNs:\n", df)
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)), columns=['A', 'B', 'C', 'D'])
df.iloc[2, 1], df.iloc[4, 3] = np.nan, np.nan  # Introduce NaNs

print("Original with NaNs:\n", df)
print("\nAfter Dropping NaNs:\n", df.dropna())
```

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 50, 10)
})
print(df)
print("\nAggregated:\n", df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 4)), columns=['A', 'B', 'C', 'D'])
df.iloc[2, 1], df.iloc[4, 3] = np.nan, np.nan  # Introduce NaNs

print("Original with NaNs:\n", df)
print("\nAfter Dropping NaNs:\n", df.dropna())
```

```python
df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], 10),
    'Value': np.random.randint(1, 50, 10)
})
print(df)
print("\nAggregated:\n", df.groupby('Category')['Value'].agg(['sum', 'mean']))
```

```python
df = pd.DataFrame({
    'Product': ['P1', 'P2', 'P3', 'P4', 'P5'],
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Home', 'Clothing'],
    'Sales': [150, 80, 200, 120, 95]
})
print(df.groupby('Category')['Sales'].sum())
```

```python
df1 = pd.DataFrame({'id': [1, 2, 3], 'name': ['Alice', 'Bob', 'Charlie']})
df2 = pd.DataFrame({'id': [1, 2, 3], 'score': [85, 90, 95]})

merged_df = pd.merge(df1, df2, on='id')
print(merged_df)
```

```python
arrays = [['A', 'A', 'B', 'B'], [1, 2, 1, 2]]
index = pd.MultiIndex.from_arrays(arrays, names=('Letter', 'Number'))
df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("Original:\n", df)
print("\nSlicing 'A':\n", df.loc['A'])
```

```python
index = pd.MultiIndex.from_tuples([('A', 'Sub1'), ('A', 'Sub2'), ('B', 'Sub1'), ('B', 'Sub2')], names=['Category', 'SubCategory'])
df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print(df)
print("\nSum by Category:\n", df.groupby('Category').sum())
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (5, 3)), columns=['A', 'B', 'C'])
print("Original:\n", df)
print("\nDoubled:\n", df.map(lambda x: x * 2))
```

```python
df = pd.DataFrame(np.random.randint(1, 10, (6, 3)), columns=['A', 'B', 'C'])
df['Sum'] = df.apply(lambda row: row.sum(), axis=1)
print(df)
```
